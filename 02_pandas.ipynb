# Topic 2 — Pandas
### Theory → tiny example → experiment.

**Why Pandas?** NumPy handles raw numeric arrays. Real data comes as **tables** — rows of observations,
columns of features, often with missing values and mixed types (numbers, text, dates). Pandas is built
on top of NumPy and adds labels, column names, and table-like operations.

- **Series** → one labeled 1D column of data
- **DataFrame** → a full table (many Series sharing an index)
- **Feature** → a column (an input variable)
- **Sample / observation** → a row (one data point)
- **Target / label** → the column you're trying to predict

```python
import pandas as pd
import numpy as np

print(pd.__version__)
```

## 1. Series and DataFrame — the two core objects

```python
# A Series: one labeled column
s = pd.Series([10, 20, 30, 40], name="score")
print(s)
print("values:", s.values)   # underlying NumPy array
print("index:", s.index)

# A DataFrame: a dict of columns, each a Series, sharing a row index
df = pd.DataFrame({
    "text": ["you are stupid", "great job today", "i hate you", "nice work"],
    "label": [1, 0, 1, 0],     # 1 = cyberbullying, 0 = not (toy example)
})
print(df)
print("shape:", df.shape)
```

## 2. Loading & inspecting data

In a real project you'll almost always start with `pd.read_csv(...)`. Colab tip:
upload a file via the left sidebar's Files panel, or mount Google Drive, then read from that path.

```python
# Example: reading a CSV (uncomment once you've uploaded a file in Colab)
# df = pd.read_csv("/content/your_dataset.csv")

# For now, build a slightly bigger toy dataset in memory
df = pd.DataFrame({
    "id": range(1, 9),
    "text": [
        "you are stupid", "great job today", "i hate you", "nice work",
        "nobody likes you", "have a good day", "you're worthless", "well done team"
    ],
    "label": [1, 0, 1, 0, 1, 0, 1, 0],
    "platform": ["insta", "insta", "twitter", "twitter", "insta", "twitter", "insta", "twitter"],
})

print(df.head())        # first 5 rows
print()
print(df.tail(3))       # last 3 rows
print()
print(df.info())        # column dtypes, non-null counts
print()
print(df.describe())    # summary stats of numeric columns
```

## 3. Selecting rows & columns

```python
# Select a single column -> returns a Series
print(df["text"].head())

# Select multiple columns -> returns a DataFrame
print(df[["text", "label"]].head())

# Select rows by position: iloc
print(df.iloc[0])        # first row
print(df.iloc[0:3])      # first 3 rows

# Select rows by label/condition: loc
print(df.loc[df["label"] == 1])   # all cyberbullying rows
```

## 4. Filtering

```python
# Boolean mask filtering — the pattern you'll use constantly
bullying_only = df[df["label"] == 1]
print(bullying_only)

# Combine conditions with & (and), | (or) — note the parentheses, required!
insta_bullying = df[(df["label"] == 1) & (df["platform"] == "insta")]
print(insta_bullying)

# Filter by text content
contains_you = df[df["text"].str.contains("you")]
print(contains_you)
```

## 5. Sorting

```python
print(df.sort_values("id", ascending=False))
print(df.sort_values(["platform", "label"]))
```

## 6. GroupBy & aggregation

This is the "split → apply → combine" pattern: split rows into groups, apply a function to each group,
combine results. Essential for exploring class balance and per-category stats.

```python
# Count rows per label -> check class balance (important for imbalanced datasets!)
print(df.groupby("label").size())

# Multiple stats at once, grouped by platform
print(df.groupby("platform")["label"].agg(["count", "mean", "sum"]))
# mean of a 0/1 label = the fraction that are cyberbullying in that group
```

## 7. Merging / joining tables

```python
users = pd.DataFrame({
    "platform": ["insta", "twitter"],
    "region": ["global", "global"],
})

merged = df.merge(users, on="platform", how="left")
print(merged.head())

# how="left"/"right"/"inner"/"outer" — same concept as SQL joins
```

## 8. Missing values & duplicates

Real datasets are messy. You'll deal with this constantly before any modeling step.

```python
messy = pd.DataFrame({
    "text": ["hello", None, "hi there", "hello"],
    "label": [0, 1, np.nan, 0],
})
print(messy)

print("\nnull counts:\n", messy.isnull().sum())

# Drop rows with any missing value
print("\ndropna:\n", messy.dropna())

# Fill missing values instead of dropping
print("\nfillna:\n", messy.fillna({"text": "unknown", "label": 0}))

# Duplicates
print("\nduplicated rows:\n", messy.duplicated())
print("\ndrop_duplicates:\n", messy.drop_duplicates())
```

## 9. Data type conversion & string operations

```python
df2 = df.copy()

# Convert dtype
df2["label"] = df2["label"].astype(int)
print(df2.dtypes)

# String ops — very relevant for your NLP preprocessing later
print(df2["text"].str.lower().head())
print(df2["text"].str.len().head())          # character count per text
print(df2["text"].str.split().head())         # split into word lists
print(df2["text"].str.replace("you", "u").head())
```

## 10. Exercise — mini EDA (exploratory data analysis) pass

Do this sequence on any dataset before modeling — it's the standard first step in every ML project,
including your cyberbullying paper.

```python
# 1. shape and dtypes
print(df.shape)
print(df.dtypes)

# 2. missing values
print(df.isnull().sum())

# 3. class balance of the target
print(df["label"].value_counts())
print(df["label"].value_counts(normalize=True))   # as proportions

# --- Try it yourself ---
# 1. Add 5 more rows to `df` with your own text/label/platform values.
# 2. Group by label and compute the average text length (hint: df["text"].str.len()).
# 3. Filter to just rows where text length > 15 characters.
```
