# Topic 1 — NumPy
### Theory → tiny example → experiment, all in one notebook.

**Why NumPy first?** Every ML/DL library (pandas, sklearn, PyTorch) stores data as arrays under the hood.
Data (a spreadsheet, an image, a sentence turned into numbers) is always represented as a **scalar, vector, matrix, or tensor**.

- **Scalar** → a single number → `5`
- **Vector** → a 1D list of numbers → `[1, 2, 3]`
- **Matrix** → a 2D grid of numbers → `[[1,2],[3,4]]`
- **Tensor** → any n-dimensional array (matrix is a 2D tensor, a color image is a 3D tensor)

NumPy's `ndarray` is how all of these are represented in code.

````python
import numpy as np

# Run this first in Colab if numpy isn't already there (it usually is):
# !pip install numpy -q
print(np.__version__)
````

## 1. Creating arrays — scalars, vectors, matrices, tensors

````python
# A scalar has 0 dimensions
scalar = np.array(5)

# A vector is a 1D array
vector = np.array([1, 2, 3])

# A matrix is a 2D array (rows x columns)
matrix = np.array([[1, 2], [3, 4]])

# A tensor is any 3D+ array — e.g. a tiny "batch of 2 grayscale 2x2 images"
tensor = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

for name, arr in [("scalar", scalar), ("vector", vector), ("matrix", matrix), ("tensor", tensor)]:
    print(f"{name}: ndim={arr.ndim}, shape={arr.shape}")
    print(arr, "\n")
````

## 2. Shape, dimensions, indexing & slicing

`.shape` tells you (rows, columns, ...). This is the single most important attribute in ML code —
almost every bug in a neural network training loop is a **shape mismatch**.

````python
X = np.array([[10, 20, 30],
              [40, 50, 60],
              [70, 80, 90]])

print("shape:", X.shape)        # (3 rows, 3 columns)
print("ndim:", X.ndim)

# Indexing: X[row, column]
print("element at row 1, col 2:", X[1, 2])   # 60

# Slicing: X[row_slice, col_slice]
print("first column:\n", X[:, 0])   # X[:, 0]  -> all rows, column 0
print("first row:\n", X[0, :])      # X[0, :]  -> row 0, all columns
print("top-left 2x2 block:\n", X[:2, :2])
````

## 3. Reshaping

Reshape changes how the *same* data is arranged, without changing the data itself.
Very common before feeding data into a model (e.g. flattening an image into a vector).

````python
a = np.arange(12)              # [0, 1, ..., 11]
print("original:", a.shape, a)

b = a.reshape(3, 4)             # 3 rows, 4 cols
print("reshaped to (3,4):\n", b)

c = a.reshape(2, 3, 2)          # a 3D tensor
print("reshaped to (2,3,2):\n", c)

# -1 means "figure this dimension out for me"
d = a.reshape(4, -1)
print("reshape(4, -1) shape:", d.shape)
````

## 4. Broadcasting & vectorization

**Vectorization** = doing an operation on the whole array at once (fast, no Python `for` loop).
**Broadcasting** = NumPy's rule for applying an operation between arrays of *different* shapes,
by "stretching" the smaller one automatically, without actually copying data.

````python
v = np.array([1, 2, 3, 4, 5])

# Vectorized (fast, no loop):
squared = v ** 2
print("squared:", squared)

# Compare: the slow, non-vectorized way (never do this in real code)
squared_loop = [x**2 for x in v]
print("same result via loop:", squared_loop)

# Broadcasting: adding a scalar to every element
print("v + 10 =", v + 10)

# Broadcasting: adding a (3,) vector to each row of a (2,3) matrix
M = np.array([[1, 2, 3], [4, 5, 6]])
row = np.array([10, 20, 30])
print("M + row =\n", M + row)   # row is "broadcast" across both rows of M
````

## 5. Element-wise operations vs aggregation

````python
a = np.array([1, 2, 3, 4])
b = np.array([10, 20, 30, 40])

# Element-wise: same-shaped arrays combined position by position
print("a + b =", a + b)
print("a * b =", a * b)     # NOT matrix multiplication, just elementwise

# Aggregation: collapse an array into a single summary number
print("sum:", a.sum())
print("mean:", a.mean())
print("max:", a.max(), " min:", a.min())
print("std:", a.std())

# Aggregation along an axis of a matrix
M = np.array([[1, 2, 3], [4, 5, 6]])
print("column sums (axis=0):", M.sum(axis=0))   # sum down each column
print("row sums (axis=1):", M.sum(axis=1))      # sum across each row
````

## 6. Random numbers

Used everywhere: initializing model weights, shuffling data, train/test splits, dropout.

````python
rng = np.random.default_rng(seed=42)   # seed = reproducible results

print("uniform [0,1):", rng.random(5))
print("normal(0,1):", rng.normal(loc=0, scale=1, size=5))
print("random ints 0-9:", rng.integers(0, 10, size=5))

# Shuffling — how you'd shuffle a dataset before splitting train/test
data = np.arange(10)
rng.shuffle(data)
print("shuffled:", data)
````

## 7. Matrix operations: dot product, transpose, matmul

These three are the mathematical backbone of linear regression and every neural network layer.

- **Dot product** of two vectors: `a·b = Σ aᵢbᵢ` → a single number.
- **Transpose**: flips a matrix over its diagonal (rows become columns).
- **Matrix multiplication**: `(n,k) @ (k,m) -> (n,m)`. The inner dimensions must match.

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

# Dot product of two vectors
print("dot product a·b =", np.dot(a, b))   # 1*4 + 2*5 + 3*6 = 32

X = np.array([[1, 2], [3, 4], [5, 6]])   # shape (3, 2)
print("X:\n", X, "\nshape:", X.shape)

# Transpose: (3,2) -> (2,3)
print("X.T:\n", X.T, "\nshape:", X.T.shape)

W = np.array([[1, 0], [0, 1], [1, 1]])   # shape (3, 2)

# Matrix multiplication needs inner dims to match: (2,3) @ (3,2) -> (2,2)
result = X.T @ W          # same as np.dot(X.T, W) or np.matmul(X.T, W)
print("X.T @ W =\n", result, "\nshape:", result.shape)
````

## 8. The "must understand" cheat-sheet

These six lines cover ~90% of what you'll type in every ML notebook you ever write:

```python
X.shape        # dimensions of the array
X[:, 0]        # all rows, column 0
X[0, :]        # row 0, all columns
X.reshape()    # rearrange without changing data
np.dot()       # dot product / matrix multiply
X.T            # transpose
```

## 9. Exercise — implement `y = Xw + b` by hand

This is the exact computation that both **linear regression** and every **neural network layer**
perform internally. Doing it once with only NumPy removes the "black box" feeling later.

````python
# 4 samples, 2 features each (like: [hours_studied, hours_slept])
X = np.array([
    [2, 9],
    [1, 5],
    [3, 6],
    [5, 1],
])

w = np.array([0.5, -0.2])   # weight for each feature
b = 1.0                     # bias

# y = Xw + b   -> matrix-vector product, then add bias (broadcasting)
y_pred = X @ w + b
print("predictions:", y_pred)

# --- Try it yourself ---
# 1. Change w and b, see how predictions shift.
# 2. Add a 3rd feature column to X and a matching 3rd weight to w.
# 3. Compute the same thing with a manual for-loop and confirm it matches X @ w + b.
````
